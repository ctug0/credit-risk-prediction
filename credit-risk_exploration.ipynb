```python
from ucimlrepo import fetch_ucirepo
import pandas as pd

credit = fetch_ucirepo(id=144)

df = credit.data.original

df.head()
```

```python
df.columns = [
    "checking_account",
    "duration",
    "credit_history",
    "purpose",
    "credit_amount",
    "savings",
    "employment",
    "installment_rate",
    "personal_status",
    "other_debtors",
    "residence_duration",
    "property",
    "age",
    "other_installment_plans",
    "housing",
    "existing_credits",
    "job",
    "dependents",
    "telephone",
    "foreign_worker",
    "Good/Bad Credit"
]

pd.set_option("display.max_columns", None)

df.head()
```

```python
data = pd.DataFrame(df)
pd.set_option("display.max_rows", 20)
data
```

```python
data.info()
```

```python
data[['duration']].describe()
```

```python
data[['Good/Bad Credit']].value_counts()
```

```python
%matplotlib inline

data.boxplot(column='duration', by ='Good/Bad Credit')

short_loans = data[data['duration'] <= 12]
short_loans['Good/Bad Credit'].value_counts()
```

```python
medium_loans = data[(data['duration'] > 12) & (data['duration'] <= 36)]
medium_loans['Good/Bad Credit'].value_counts()
```

```python
long_loans = data[data['duration'] > 36]
long_loans['Good/Bad Credit'].value_counts()
```
